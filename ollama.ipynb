```python
import requests
import json
```

```python
from langchain_ollama import OllamaLLM
from langchain_core.prompts import ChatPromptTemplate
```

```python
template = """
give me {number} parallel sentences in english and polish.

context: video game dialogues 
"""
```

```python
model = OllamaLLM(model="gemma")
prompt = ChatPromptTemplate.from_template(template)
chain = prompt | model

result = chain.invoke({"number":"5"})
print(result)
```

```python
ollama_api_url = "http://localhost:11434/api/generate"
```

```python
def query_ollama(prompt):
    headers = {"Content-Type": "application/json"}
    data = {"model": "llama2", "prompt": prompt}
    response = requests.post(ollama_api_url, headers=headers, data=json.dumps(data))
    if response.status_code == 200:
        return response.json()["choices"][0]["text"].strip()
    else:
        print(f"Error querying Ollama: {response.text}")
        return None
```

```python
prompt = "What is the capital of France?"
answer = query_ollama(prompt)
print(answer)
```
